```python
import functools

# Decorador personalizado para loguear las operaciones
def log_operacion(operacion):
    def decorador(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            print(f"\n>>> Iniciando operación: {operacion}")
            resultado = func(*args, **kwargs)
            print(f">>> Operación '{operacion}' finalizada exitosamente.")
            return resultado
        return wrapper
    return decorador


class Mascota:
    """Clase principal para representar una Mascota en el sistema."""

    def __init__(self, numero_ficha, nombre, especie, peso_kg):
        self._numero_ficha = numero_ficha
        self.nombre = nombre
        self.especie = especie
        self.peso_kg = peso_kg  # pasa por el setter

    @property
    def numero_ficha(self):
        return self._numero_ficha

    @property
    def peso_kg(self):
        return self._peso

    @peso_kg.setter
    def peso_kg(self, valor):
        if valor <= 0:
            raise ValueError("El peso debe ser mayor a 0.")
        self._peso = valor

    def __str__(self):
        return f"[Ficha: {self._numero_ficha}] {self.nombre} - {self.especie} | {self.peso_kg} kg"


class Veterinaria:
    """Clase para manejar el ABM de mascotas."""

    def __init__(self):
        self.mascotas = {}

    @log_operacion("Alta de Mascota")
    def agregar_mascota(self, numero_ficha, nombre, especie, peso_kg):
        if numero_ficha in self.mascotas:
            print(f"Error: Ya existe una mascota con ficha {numero_ficha}.")
            return False

        try:
            nueva = Mascota(numero_ficha, nombre, especie, peso_kg)
            self.mascotas[numero_ficha] = nueva
            print(f"Mascota agregada: {nueva.nombre}")
            return True
        except ValueError as e:
            print(f"Error de validación: {e}")
            return False

    @log_operacion("Listado de Mascotas")
    def listar_mascotas(self):
        if not self.mascotas:
            print("No hay mascotas registradas.")
            return

        print("\n--- Lista de Mascotas ---")
        for mascota in self.mascotas.values():
            print(mascota)
        print("--------------------------")

    @log_operacion("Modificación de Mascota")
    def actualizar_mascota(self, numero_ficha, nombre=None, especie=None, peso_kg=None):
        if numero_ficha not in self.mascotas:
            print("Error: Mascota no encontrada.")
            return False

        mascota = self.mascotas[numero_ficha]

        try:
            if nombre:
                mascota.nombre = nombre
            if especie:
                mascota.especie = especie
            if peso_kg is not None:
                mascota.peso_kg = peso_kg
            print("Mascota actualizada correctamente.")
            return True
        except ValueError as e:
            print(f"Error de validación: {e}")
            return False

    @log_operacion("Baja de Mascota")
    def eliminar_mascota(self, numero_ficha):
        if numero_ficha in self.mascotas:
            mascota = self.mascotas.pop(numero_ficha)
            print(f"Mascota eliminada: {mascota.nombre}")
            return True
        else:
            print("Error: Mascota no encontrada.")
            return False

    @log_operacion("Listado por Especie")
    def listar_por_especie(self, especie_buscada):
        encontrados = [m for m in self.mascotas.values() if m.especie.lower() == especie_buscada.lower()]

        if not encontrados:
            print(f"No se encontraron mascotas de especie '{especie_buscada}'.")
            return

        print(f"\n--- Mascotas tipo {especie_buscada} ---")
        for m in encontrados:
            print(m)
        print("-------------------------------------")


def mostrar_menu():
    print("\n" + "="*40)
    print(" SISTEMA VETERINARIA ")
    print("="*40)
    print("[1] Agregar Mascota")
    print("[2] Listar Mascotas")
    print("[3] Actualizar Mascota")
    print("[4] Eliminar Mascota")
    print("[5] Listar por Especie")
    print("[6] Salir")
    print("="*40)


def main():
    vet = Veterinaria()

    # Datos de prueba
    vet.agregar_mascota(1, "Firulais", "Perro", 12.5)
    vet.agregar_mascota(2, "Michi", "Gato", 4.2)

    while True:
        mostrar_menu()
        opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            try:
                ficha = int(input("Número de ficha: "))
                nombre = input("Nombre: ")
                especie = input("Especie: ")
                peso = float(input("Peso (kg): "))
                vet.agregar_mascota(ficha, nombre, especie, peso)
            except ValueError:
                print("Datos inválidos.")

        elif opcion == "2":
            vet.listar_mascotas()

        elif opcion == "3":
            try:
                ficha = int(input("Ficha a modificar: "))
                nombre = input("Nuevo nombre: ").strip() or None
                especie = input("Nueva especie: ").strip() or None
                peso_str = input("Nuevo peso: ").strip()
                peso = float(peso_str) if peso_str else None
                vet.actualizar_mascota(ficha, nombre, especie, peso)
            except ValueError:
                print("Datos inválidos.")

        elif opcion == "4":
            try:
                ficha = int(input("Ficha a eliminar: "))
                vet.eliminar_mascota(ficha)
            except ValueError:
                print("Ficha inválida.")

        elif opcion == "5":
            especie = input("Ingrese especie a buscar: ")
            vet.listar_por_especie(especie)

        elif opcion == "6":
            print("Saliendo...")
            break

        else:
            print("Opción inválida.")


if __name__ == "__main__":
    main()
```
